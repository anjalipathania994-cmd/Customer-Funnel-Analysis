```python

import matplotlib.pyplot as plt

# Revenue by each channel
df.groupby("Channel")["Revenue"].sum().plot(
    kind="bar",color = "coral",
    title="Revenue by Channel"
)

plt.xlabel("Channel")
plt.ylabel("Total Revenue")
plt.tight_layout()

plt.savefig("2_channel.png")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Revenue by product category
df.groupby("Product_Category")["Revenue"].sum().plot(
    kind="bar",color = "green",
    title="Revenue by Category"
)

plt.xlabel("Product Category")
plt.ylabel("Total Revenue")
plt.tight_layout()

plt.savefig("3_category.png")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Users count by each region
region_counts = df['Region'].value_counts()

region_counts.plot(
    kind='bar',color =["purple","red","blue","coral","orange"],
    title="Users by Region"
)

plt.xlabel("Region")
plt.ylabel("Number of Users")
plt.tight_layout()

plt.show()
```

```python
# calculates unique users at each event stage, then computes percentage change between stages to show drop-off
funnel = df.groupby('Event')['User_ID'].nunique()

drop_off = funnel.pct_change() * 100
drop_off
```

```python
import matplotlib.pyplot as plt

# Bounce flag distribution
df["Bounce_Flag"].value_counts().plot(
    kind="bar",color = ["red","green"],
    title="Bounce Flag"
)

plt.xlabel("Bounce Flag")
plt.ylabel("Count")
plt.tight_layout()

plt.savefig("5_bounce.png")
plt.show()
```

```python
# Convert timestamp to datetime format
df['timestamp'] = pd.to_datetime(df['timestamp'])

df = df.set_index('timestamp')
```

```python
#daily revenue trend
daily_revenue = df['Revenue'].resample('D').sum()

# Plot
plt.figure(figsize=(12,6))
plt.plot(daily_revenue)
plt.title("Daily Revenue Trend")
plt.xlabel("Date")
plt.ylabel("Revenue")
plt.show()
```

```python
#weekly revenue trend
weekly_revenue = df['Revenue'].resample('W').sum()

plt.figure(figsize=(10,5))
plt.plot(weekly_revenue)
plt.title("Weekly Revenue Trend")
plt.xlabel("Week")
plt.ylabel("Revenue")
plt.show()
```

```python
#daily active user trend
daily_users = df.resample('D')['User_ID'].nunique()

plt.figure(figsize=(12,5))
plt.plot(daily_users)
plt.title("Daily Active Users Trend")
plt.xlabel("Date")
plt.ylabel("Unique Users")
plt.show()
```



## 📊 Customer Funnel Analysis - EDA Summary

After performing data cleaning and exploratory data analysis, several key patterns were identified in the customer funnel. A significant drop-off was observed in the later stages, particularly from Signup to Purchase, indicating friction in the conversion process. Conversion rates decline at each stage of the funnel, highlighting inefficiencies in user progression.

Mobile devices contribute a large share of total traffic but show relatively lower conversion rates compared to desktop users. User activity varies over time, with higher engagement observed during specific periods such as weekends. Additionally, returning users demonstrate higher conversion rates than new users, indicating strong retention behavior.

---

## 📈 Key Insights

- The highest drop-off occurs at the final stage, suggesting issues in the checkout process.

- Conversion rates decrease at each funnel stage, indicating a need to improve user flow.

- **Mobile users** have lower conversion rates despite high traffic, pointing to poor mobile experience.

- **Returning users** convert better than new users, highlighting strong customer retention.

- User engagement is higher during **weekends**, indicating peak activity periods.

- Revenue is concentrated in specific regions, suggesting dependency on limited markets.

## ✅ Recommendations

- Optimize and simplify the checkout process to reduce drop-offs at the final stage.

- Improve mobile user experience to increase mobile conversion rates.

- Streamline the overall user journey to enhance conversion across funnel stages.

- Focus on retaining existing users while improving strategies to convert new users.

- Run targeted marketing campaigns during high-engagement periods (e.g., weekends).

- Expand marketing efforts in underperforming regions to balance revenue distribution.

---

## 📌 Conclusion

The analysis highlights critical bottlenecks in the customer funnel, particularly in conversion and user experience. By addressing these issues and leveraging identified trends, businesses can improve conversion rates, enhance user engagement, and drive overall growth.
